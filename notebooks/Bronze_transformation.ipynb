```python
customers_df = spark.table("ecommerce_project.project.customers")
products_df = spark.table("ecommerce_project.project.products")
orders_df = spark.table("ecommerce_project.project.orders")
payments_df = spark.table("ecommerce_project.project.payments")
deliveries_df = spark.table("ecommerce_project.project.deliveries")
```

```python
customers_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("ecommerce_project.project.bronze_customers")
```

```python
products_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("ecommerce_project.project.bronze_products")

orders_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("ecommerce_project.project.bronze_orders")

payments_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("ecommerce_project.project.bronze_payments")

deliveries_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("ecommerce_project.project.bronze_deliveries")
```
